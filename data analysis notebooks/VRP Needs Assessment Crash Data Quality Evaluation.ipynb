# Crash Table Linked Address Fill Percentage

```python
Parks = sorted(crashcountfiltered['Park'].unique())
Years = sorted(crashcountfiltered['YEAR'].unique())
complete = []

for i in Parks:
    for j in Years:
        df = crashcountfiltered.loc[(crashcountfiltered['Park']==i)&(crashcountfiltered['YEAR']==j)]
        if df.empty:
            complete.append('N/A')
        else:    
            notNull = df.loc[df['Linked_Address'].notnull()]
            complete.append(np.nan_to_num(round(notNull.shape[0]/df.shape[0]*100,2)))
```

# Lat/Long Within Park Boundary Percentage

```python
import geopandas as gpd
notNull = crashcountfiltered.loc[(crashcountfiltered['LATITUDE'].notnull())&(crashcountfiltered['LONGITUDE'].notnull())]
ALLCRASHES_ALLCOORDS = notNull.loc[(notNull['LATITUDE'].abs() > 0)&(notNull['LONGITUDE'].abs() > 0)]
ALLBOUNDARIES=gpd.read_file("./shapefiles/nps_boundary.shp")
    
#This is the list of parks that (1) are not assigned the NACA label in the NPS Lands Boundary and
#(2) also have crashes with a park and coordinate filled; may not be entirely complete but should cover
#any parks that would be in the top 15 for the NCR
    
nonNACAlist=["SUIT","MLND","NACE","NACC","NAMA","BAWA","GWMP","ROCR","LINC","MANA","JEFM","CATO",
            "PRWI","CLBA","THIS","COGA","FRDE","GREE","ANAC","WASH","ANTI","PAAV","PISC","MONO",
            "WHHO","FOWA","HAFE","ARHO","WOTR","MLKM","DDEM","CHOH","JOFK","BEPA","LYBA","MALL",
            "MABE","WWII","FOTH","WAMO","KOWA","FRDO","CAWO","VIVE","WWIM","THJE","CHND","WRWR"]
    
for crash in range(len(ALLCRASHES_ALLCOORDS)):
    park=ALLCRASHES_ALLCOORDS.iloc[crash]['Park']
    region=ALLCRASHES_ALLCOORDS.iloc[crash]['RGN']
        
    if park not in nonNACAlist and region=="NCR":
        ALLCRASHES_ALLCOORDS['Park'][crash] = "NACA"

Parks = sorted(ALLCRASHES_ALLCOORDS['Park'].unique())
Years = sorted(crashcountfiltered['YEAR'].unique())
complete = []

for i in Parks:
    for j in Years:
        df = ALLCRASHES_ALLCOORDS.loc[(ALLCRASHES_ALLCOORDS['Park']==i)&(ALLCRASHES_ALLCOORDS['YEAR']==j)]
        if df.empty:
            complete.append('N/A')
        else:    
            park_crashes=gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df['LONGITUDE'],df['LATITUDE']))
            proj_park_crashes=park_crashes.set_crs(epsg=3857)
            park_polygon=ALLBOUNDARIES.loc[ALLBOUNDARIES['UNIT_CODE']==i]
            proj_park_polygon=park_polygon.set_crs(epsg=3857,allow_override=True)
            inBoundary = len(gpd.sjoin(proj_park_polygon,proj_park_crashes,how='left'))
            complete.append(np.nan_to_num(round(inBoundary/df.shape[0]*100,2)))
```

```python
complete
```

```python
Filled = pd.DataFrame({'Park':np.repeat(Parks, len(Years)), 'Year':Years*len(Parks), 'NonZero LatLong Within Park Boundary Percentage':complete})
Filled.to_csv("./VRP Needs Assessment/NonZero LatLong Within Park Boundary Percentages.csv", index=False)
```

# Injury/Fatality Indicator Matches Injury Severity

```python
crash_details_crashcountfiltered_filled = crash_details_crashcountfiltered.loc[crash_details_crashcountfiltered['Injury_or_Fatal_Crash'].notnull()]
passenger_crashcountfiltered_slim = passenger_crashcountfiltered[['IMARS_Record_No']]
passenger_crashcountfiltered_slim['No Injury']= np.where(passenger_crashcountfiltered['Injury_Severity']=='01. No injury', 1,0)
passenger_crashcountfiltered_slim['Possible Injury']= np.where(passenger_crashcountfiltered['Injury_Severity']=='02. Possible injury', 1,0)
passenger_crashcountfiltered_slim['Non-incapacitating Injury']= np.where(passenger_crashcountfiltered['Injury_Severity']=='03. Non-incapacitating injury', 1,0)
passenger_crashcountfiltered_slim['Incapacitating Injury']= np.where(passenger_crashcountfiltered['Injury_Severity']=='04. Incapacitating injury', 1,0)
passenger_crashcountfiltered_slim['Fatality']= np.where(passenger_crashcountfiltered['Injury_Severity']=='05. Fatal', 1,0)
passenger_crashcountfiltered_slim_agg = passenger_crashcountfiltered_slim.groupby(by=['IMARS_Record_No']).sum().reset_index()
passenger_crashcountfiltered_slim_agg['Crash_Severity'] = 0 # initialize column with dummy data to replace with for loop
length = range(passenger_crashcountfiltered_slim_agg.shape[0]) # search through all rows in aggregated dataset (to examine all injuries from each crash as a whole)
for i in length: # determine most severe injury outcome
    if passenger_crashcountfiltered_slim_agg['Fatality'].iloc[i] > 0:
        passenger_crashcountfiltered_slim_agg['Crash_Severity'].iloc[i] = 'Fatal'
    elif passenger_crashcountfiltered_slim_agg['Incapacitating Injury'].iloc[i] > 0:
        passenger_crashcountfiltered_slim_agg['Crash_Severity'].iloc[i] = 'Incap'
    elif passenger_crashcountfiltered_slim_agg['Non-incapacitating Injury'].iloc[i] > 0:
        passenger_crashcountfiltered_slim_agg['Crash_Severity'].iloc[i] = 'Non-Incap'
    elif passenger_crashcountfiltered_slim_agg['Possible Injury'].iloc[i] > 0:
        passenger_crashcountfiltered_slim_agg['Crash_Severity'].iloc[i] = 'Possible'
    else:
        passenger_crashcountfiltered_slim_agg['Crash_Severity'].iloc[i] = 'No Inj'
```
