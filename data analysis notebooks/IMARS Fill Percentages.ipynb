# Determine Any Fill and Non-Unknown Fill Percentages

## Raw Reports

### Crash Table

```python
crashcolumns = imars_crash.columns
filled = []
filledNotUnknownPerc = []

for i in crashcolumns:
    notNull = imars_crash.loc[imars_crash[i].notnull()]
    # determine fill percentages (counting "unknowns")
    full = np.nan_to_num(round(notNull.shape[0]/imars_crash.shape[0]*100,0)).astype(int) 
    filled.append(full)

    # drop reports on "null island", which is the lat/long equivalent of "unknown" to determine percentage of remaining reports filled with REAL lat/long data
    if i=="Latitude":
        fullNotUnknown = notNull.loc[notNull[i] != 0] 
        fullNotUnknownPerc = np.nan_to_num(round(fullNotUnknown.shape[0]/imars_crash.shape[0]*100,0)).astype(int)
        filledNotUnknownPerc.append(fullNotUnknownPerc)
    elif i=="Longitude":
        fullNotUnknown = notNull.loc[notNull[i] != 0] 
        fullNotUnknownPerc = np.nan_to_num(round(fullNotUnknown.shape[0]/imars_crash.shape[0]*100,0)).astype(int)
        filledNotUnknownPerc.append(fullNotUnknownPerc)
    else:
        fullNotUnknownPerc = 'NA'
        filledNotUnknownPerc.append(fullNotUnknownPerc)
        
filledNotUnknownPerc
```

```python
Filled = {'Column':crashcolumns, 'Fill Percentage':filled, 'Non-Unknown Fill Percentage':filledNotUnknownPerc}
Filled_CrashTable = pd.DataFrame(Filled)
Filled_CrashTable.to_csv("IMARS Crash Table Fill Percentages.csv", index=False)
```

### Passenger Table

```python
passengercolumns = imars_passenger.columns
filled = []
filledNotUnknownPerc = []

for i in passengercolumns:
    notNull = imars_passenger.loc[imars_passenger[i].notnull()]
    # determine fill percentages (including "unknowns")
    full = np.nan_to_num(round(notNull.shape[0]/imars_passenger.shape[0]*100,0)).astype(int)
    filled.append(full)

    # determine percent of reports where the crash reporter entered something OTHER THAN the unknown attributes listed 
        # (drop reports with these exact values, but keep reports where unknown was one of many REAL data attributes selected: e.g., "20. Avoiding an object on road; 24. Over-correcting/over-steering; 99. Unknown")
    fullNotUnknown = notNull.loc[notNull[i].isin(["99. Unknown", "06. Unknown if distracted", "07. Unknown if distracted"])==False] 
    fullNotUnknownPerc = np.nan_to_num(round(fullNotUnknown.shape[0]/imars_passenger.shape[0]*100,0)).astype(int)
    filledNotUnknownPerc.append(fullNotUnknownPerc)
filledNotUnknownPerc
```

```python
Filled = {'Column':passengercolumns, 'Fill Percentage':filled, 'Non-Unknown Fill Percentage':filledNotUnknownPerc}
Filled_PassengerTable = pd.DataFrame(Filled)
Filled_PassengerTable.to_csv("IMARS Passenger Table Fill Percentages.csv", index=False)
```

### Crash Table

```python
# drop reports with missing or duplicate record numbers
imars_crash = imars_crash.dropna(subset=['IMARS_Record_No'])
imars_crash = imars_crash.drop_duplicates(subset=['IMARS_Record_No'])
imars_crash.shape
```

```python
# pull out crashes that already have park names
imars_crash_parks = imars_crash.dropna(subset=['Park'])
imars_crash_parks.shape
```

```python
# revert flipped lat/long
need_revert = imars_crash['Latitude'].abs() > 70
imars_crash.loc[need_revert, ['Latitude', 'Longitude']] = (imars_crash.loc[need_revert, ['Longitude', 'Latitude']].values)
```

```python
# adjust coordinates to correct hemisphere
imars_crash['Latitude'].loc[imars_crash['Latitude'] < 0] = imars_crash['Latitude'].loc[imars_crash['Latitude'] < 0]*(-1)
imars_crash['Longitude'].loc[imars_crash['Longitude'] > 0] = imars_crash['Longitude'].loc[imars_crash['Longitude'] > 0]*(-1)
imars_crash['Longitude'].loc[imars_crash['Longitude'].abs() > 180] = imars_crash['Longitude'].loc[imars_crash['Longitude'].abs() > 180]/10
imars_crash['Latitude'].loc[imars_crash['Latitude'].abs() > 80] = imars_crash['Latitude'].loc[imars_crash['Latitude'].abs() > 80]/10
```

```python
# fill in missing park names via spatial join with park boundaries for those with lat/longs
import geopandas as gpd
imars_crash_coords = imars_crash.loc[imars_crash['Park'].isnull()].dropna(subset=['Latitude','Longitude'])
imars_crash_coords_geo = gpd.GeoDataFrame(imars_crash_coords, geometry=gpd.points_from_xy(imars_crash_coords.Longitude, imars_crash_coords.Latitude))
filename = "./shapefiles/NPS_-_Land_Resources_Division_Boundary_and_Tract_Data_Service.geojson"
file = open(filename)
parks = gpd.read_file(file)
imars_crash_coords_geo.crs = "EPSG:4326"
parks = parks.to_crs(epsg=4326)
parks['geometry'] = parks['geometry'].buffer(0.01)
imars_crash_coords_geo_withparknames = gpd.sjoin(imars_crash_coords_geo,parks,how="left", predicate='intersects')
imars_crash_coords_geo_withparknames2 = imars_crash_coords_geo_withparknames.drop_duplicates(subset=['IMARS_Record_No'])
imars_crash_withparknames = pd.DataFrame(imars_crash_coords_geo_withparknames2.drop(columns='geometry'))
imars_crash_withparknames['Park']= imars_crash_withparknames['UNIT_CODE']
imars_crash_withparknames_NoDupsorNulls = imars_crash_withparknames.dropna(subset=['Park'])
imars_crash_expanded = pd.concat([imars_crash_parks,imars_crash_withparknames_NoDupsorNulls])
imars_crash_expanded.shape
```

```python
# pull out crashes WITHOUT park names or lat/long AND WITH location info that could be used to manually identify park name
imars_crash_noparks = imars_crash.loc[imars_crash['Park'].isnull()]
imars_crash_noparks_nocoords = imars_crash_noparks.loc[imars_crash_noparks['Latitude'].isnull()]
mask = imars_crash_noparks_nocoords.loc[:,['Linked_Common_Name','Linked_Address','NEAR_route_street_road_name','At_Intersecting_route_street_road_name']].notnull()
imars_noparks_nocoords_someinfo = imars_crash_noparks_nocoords.loc[mask.any(axis=1)]
#imars_noparks_nocoords_someinfo.to_csv("./IMARS_noparks_nocoords_someinfo.csv",index=False)
```

```python
# derive time field for assigning known lighting condition to unknown/null records
imars_crash_expanded['Crash_Date_Time'] = pd.to_datetime(imars_crash_expanded['Crash_Date_Time'], format = '%Y%m%d %H:%M:%S:%f')
imars_crash_expanded['TIME'] = imars_crash_expanded['Crash_Date_Time'].dt.strftime('%H:%M')
```

```python
# determine empty and unknown lighting fields that have the potential to be filled in
imars_crash_details_expanded['UnknownLit'] = np.where(imars_crash_details_expanded['Lighting']=="99. Unknown", 1,0)
imars_crash_details_expanded['NullLit'] = np.where(imars_crash_details_expanded['Lighting'].isnull(),1,0)
imars_crash_details_expanded['UnknownNullLit'] = imars_crash_details_expanded['UnknownLit']+imars_crash_details_expanded['NullLit']
# check sum
imars_crash_details_expanded['UnknownNullLit'].sum(), 622+556
```

```python
# replace unspecified null Lighting attributes with more accurate data based on general sun position with respect to time
# times from NOAA: https://vividmaps.com/sunrise-time-sunset-time-and-hours-of/ and https://www.google.com/search?sca_esv=572573644&q=sunrise+and+sunset+on+winter+solstice+map&tbm=isch&source=lnms&sa=X&ved=2ahUKEwjgnaaU9O6BAxUlFFkFHeLJDn8Q0pQJegQIDRAB&biw=1912&bih=966&dpr=1

length = range(imars_crash_details_expanded.shape[0]) # search through all rows in dataset
for i in length: # assign lighting condition based on crash hour when unknown
    if imars_crash_details_expanded["RGN"].iloc[i] == "NCR": # NCR has more precise guaranteed lighting times due to concentrated crash locations
        if imars_crash_details_expanded['UnknownNullLit'].iloc[i] == 1:
            if imars_crash_details_expanded['TIME'].iloc[i] > '08:29' and imars_crash_details_expanded['TIME'].iloc[i] < '15:31': # guaranteed daylight period w/ hour buffer
                imars_crash_details_expanded['Lighting'].iloc[i] = '01. Daylight'
            elif imars_crash_details_expanded['TIME'].iloc[i] > '21:59' or imars_crash_details_expanded['TIME'].iloc[i] < '04:31': # guaranteed dark period w/ hour buffer 
                imars_crash_details_expanded['Lighting'].iloc[i] = '06. Dark - unknown lighting'
    else: # all other regions with less precise guaranteed lighting times due to spread crash locations
        if imars_crash_details_expanded['UnknownNullLit'].iloc[i] == 1:
            if imars_crash_details_expanded['TIME'].iloc[i] > '09:59' and imars_crash_details_expanded['TIME'].iloc[i] < '14:31':
                imars_crash_details_expanded['Lighting'].iloc[i] = '01. Daylight'
            elif imars_crash_details_expanded['TIME'].iloc[i] > '22:59' or imars_crash_details_expanded['TIME'].iloc[i] < '03:31':
                imars_crash_details_expanded['Lighting'].iloc[i] = '06. Dark - unknown lighting' 
                
# new totals after reassignment
imars_crash_details_expanded['Lighting'].value_counts()
```

```python
minimumviablecolumns_crash = ['Linked_Common_Name','NEAR_route_street_road_name','At_Intersecting_route_street_road_name','Latitude','Longitude','Region','Park']
filled = []
filledNotUnknownPerc = []

for i in minimumviablecolumns_crash:
    notNull = imars_crash_fatalonly.loc[imars_crash_fatalonly[i].notnull()]
    # determine fill percentages (counting "unknowns")
    full = np.nan_to_num(round(notNull.shape[0]/imars_crash_fatalonly.shape[0]*100,0)).astype(int) 
    filled.append(full)

    # drop reports on "null island", which is the lat/long equivalent of "unknown" to determine percentage of remaining reports filled with REAL lat/long data
    if i=="Latitude":
        fullNotUnknown = notNull.loc[notNull[i] != 0] 
        fullNotUnknownPerc = np.nan_to_num(round(fullNotUnknown.shape[0]/imars_crash_fatalonly.shape[0]*100,0)).astype(int)
        filledNotUnknownPerc.append(fullNotUnknownPerc)
    elif i=="Longitude":
        fullNotUnknown = notNull.loc[notNull[i] != 0] 
        fullNotUnknownPerc = np.nan_to_num(round(fullNotUnknown.shape[0]/imars_crash_fatalonly.shape[0]*100,0)).astype(int)
        filledNotUnknownPerc.append(fullNotUnknownPerc)
    else: 
        fullNotUnknownPerc = 'NA'
        filledNotUnknownPerc.append(fullNotUnknownPerc)
        
filledNotUnknownPerc
```
